- Dataset300 cases 0201-0300 are treated as the nominal increment.

## Takeaways

- Dataset300 is an image-level superset: its first 200 images are the public training images, and cases 0201-0300 are the nominal increment.
- The public target JSON and Dataset300 CVJsonStd annotations are different representations and must be normalized before joint use.
- The new 100 cases add mostly datasheet-domain supervision, so they help targeted perception tasks but do not broaden every domain equally.
- Public images must not be appended as another independent corpus; merge complementary labels by image hash and keep duplicate groups split-safe.
